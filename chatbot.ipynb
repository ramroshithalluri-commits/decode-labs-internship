```python
RESPONSES = {
    # Greetings
    'hello'        : 'Hello! I am DecodBot 🤖. How can I assist you today?',
    'hi'           : 'Hi there! Ready to help. What is on your mind?',
    'hey'          : 'Hey! DecodBot online. What do you need?',

    # Identity
    'who are you'  : 'I am DecodBot — a rule-based AI built at DecodeLabs, Batch 2026.',
    'what are you' : 'I am a deterministic chatbot. No hallucinations. Pure logic. 💡',
    'your name'    : 'My name is DecodBot. Nice to meet you!',

    # DecodeLabs
    'what is decodelabs' : 'DecodeLabs is an AI training organization based in Greater Lucknow, India.',
    'decodelabs'         : 'DecodeLabs — powering the next generation of AI engineers! 🚀',

    # AI Concepts
    'what is ai'               : 'AI (Artificial Intelligence) is the simulation of human intelligence by machines.',
    'what is machine learning' : 'ML is a subset of AI where systems learn from data to improve over time.',
    'what is a chatbot'        : 'A chatbot simulates conversation with humans. You are talking to one! 🤖',
    'rule based ai'            : 'Rule-based AI uses explicit logic. 100% deterministic — no hallucinations.',

    # Help
    'help' : "I can answer: 'what is ai', 'who are you', 'joke', 'ipo model', 'what is project 1'.",

    # Fun
    'joke'      : 'Why do programmers prefer dark mode? Because light attracts bugs! 🐛',
    'how are you': 'Running at O(1) lookup speed. Never been better! ⚡',
    'thank you' : 'You are welcome! Keep building. 💪',
    'thanks'    : 'Anytime! That is what I am here for.',

    # Project
    'what is project 1' : 'Project 1 is the Rule-Based Chatbot — the foundation milestone of the DecodeLabs internship!',
    'ipo model'         : 'IPO = Input → Process → Output. Sanitize → Lookup → Respond.',
}

EXIT_COMMANDS = {'exit', 'quit', 'bye', 'goodbye', 'stop'}
FALLBACK = "❓ I don't understand that yet. Type 'help' to see what I can do."

print('✅ Knowledge Base loaded —', len(RESPONSES), 'intents ready.')
```

---
## 🔧 Phase 2 — Sanitization & Response Functions

```python
def sanitize(raw: str) -> str:
    """Phase 1 — Normalize input: lowercase + strip whitespace."""
    return raw.lower().strip()

def get_response(clean_input: str) -> str:
    """Phase 2 — O(1) dictionary lookup with fallback."""
    return RESPONSES.get(clean_input, FALLBACK)

print('✅ Functions defined.')
```

---
## 💬 Phase 3 — Run the Chatbot (The Heartbeat Loop)
> Type your message and press **Enter**. Type `exit` to quit.

```python
print('='*55)
print('   DecodeLabs · Project 1 · Batch 2026')
print('   Rule-Based AI Chatbot  🤖')
print("   Type 'help' for commands | 'exit' to quit")
print('='*55)
print()

while True:
    raw_input_text = input('You: ')
    clean_input = sanitize(raw_input_text)

    if clean_input in EXIT_COMMANDS:
        print('Bot: Goodbye! Keep coding and keep building. 🚀')
        break

    if not clean_input:
        continue

    reply = get_response(clean_input)
    print(f'Bot: {reply}\n')
```
